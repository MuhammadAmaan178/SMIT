```python
class Person():
    def __init__(self,name,age,country):
        self.name = name
        self.age = age
        self.country = country
    def display(self):
        print(f"""
        Name : {self.name}
        Age : {self.age}
        Country : {self.country}
        """)
```

```python
p1 = Person("Amaan",18,"Pakistan")
p2 = Person("Umer",21,"Australia")
p1.display()
p2.display()
```

```python
class Rectangle():
    def __init__(self,width,height):
        self.width = width
        self.height = height
    def area(self):
        return self.width*self.height
    def perimeter(self):
        return 2*(self.height+self.width)
```

```python
r1 = Rectangle(2,5)
print(r1.area())
print(r1.perimeter())
```

```python
class BankAccount():
    def __init__(self,acc_num,balance):
        self.__acc_num = acc_num
        self.__balance = balance
    def withdraw(self,acc_num,money):
        if acc_num == self.__acc_num:
            if money <= self.__balance:
                self.__balance = self.__balance - money
                return f"{money} successfully withdraw...Your balance is {self.__balance}"
            else:
                return "Insuffiecient Balance"
        else:
            return "Incorrect Account Number"
    def deposit(self,acc_num,money):
        if acc_num == self.__acc_num:
            self.__balance = self.__balance + money
            return f"{money} successfully deposit...Your balance is {self.__balance}"
        else:
            return "Incorrect Account Number"
```

```python
b1 = BankAccount(12345,999)
b1.deposit(12345,1)
```

```python
b1.withdraw(12345,1)
```

```python
class Shape():
    def area():
        pass
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return 3.145*(self.radius*self.radius)
class Triangle(Shape):
    def __init__(self,base,height):
        self.base = base
        self.height = height
    def area(self):
        return (self.base * self.height)/2
```

```python
class Bank():
    def __init__(self,names):
        self.names = names
        self.branches = []
    def add_branch(self,branch):
        if branch not in self.branches:
            self.branches.append(branch)
            return f"{branch} branch added to {self.names}"
        else:
            return f"{branch} branch already exsists"
    def remove_branch(self,branch):
        if branch in self.branches:
            self.branches.remove(branch)
            return f"{branch} branch remove from {self.names}"
        else:
            return f"{branch} branch doesnot exsists"
    def show_branches(self):
        for branch in self.branches:
            print(branch)
        
```

```python
class Product():
    def __init__(self,ID,name,price,quantity):
        self.ID = ID
        self.name = name
        self.price = price
        self.quantity = quantity
    def total_price(self):
        return f"TOTAL PRICE : {self.price * self.quantity}"
```

```python
class PersonalCareProduct(Product):
    def __init__(self,ID,name,price,quantity,warrenty_period):
        super().__init__(ID,name,price,quantity)
        self.warrenty_period  = warrenty_period
    def total_price(self):
        return f"""
        TOTAL PRICE : {self.price * self.quantity}
        WARRENTY PERIOD : {self.warrenty_period}""" 
```

```python
product1 = PersonalCareProduct(143,"Laptop",180000,1,"10 Years")
print(product1.total_price())
```

```python
class BankAccount():
    def __init__(self,acc_num,acc_name,balance):
        self.acc_num = acc_num
        self.acc_name = acc_name
        self.balance = balance
        if balance >= 0:
            All_Accounts[self.acc_num] = self.balance
        else:
            All_Accounts[self.acc_num] = 0

    def check_info(self):
        return f"""
            Account Number : {self.acc_num}
            Account Name : {self.acc_name}
            Account Balance : {self.balance}
            """
    def deposit(self,balance):
        if balance >= 0:
            self.balance = self.balance + balance
            return "Money Added Successfully"
        else:
            return "Insufficient Balance"
    def withdraw(self,balance):
        if balance <= self.balance:
            self.balance = self.balance - balance
            return "Money Withdraw Successfully"
        else:
            return "Insufficient Balance"
    def transfer(self,acc_num,balance):
        if self.acc_num != acc_num:
            if acc_num in All_Accounts:
                temp = All_Accounts[acc_num]
                temp = temp + balance
                All_Accounts[acc_num] = temp
                temp = All_Accounts[self.acc_num] 
                temp = temp - balance
                All_Accounts[self.acc_num] = temp
                return "Transistion Succesfully Done"
            else:
                return "Your Account doesnot exists"
                
```
